```python
import numpy as np
```

```python
v1 = np.array([1, 1, 0, 0], dtype=float)
v1 /= np.linalg.norm(v1)
P1 = np.outer(v1, v1) / np.dot(v1, v1)
v1
```

```python
v2 = np.array([0, 0, 1, -1], dtype=float)
v2 /= np.linalg.norm(v2)
P2 = np.outer(v2, v2) / np.dot(v2, v2)
v2
```

```python
P = P1 + P2

u, s, vt = np.linalg.svd(P)

u[:, :2].T @ v1
```

```python
u[:, :2].T @ v2
```

```python
u[:, 2:].T @ v1
```
